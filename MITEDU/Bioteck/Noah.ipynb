```python
import py3Dmol

# Replace with downloaded AlphaFold PDB file
#with open("AF-Q9UPP1-F1-model_v4.pdb", "r") as f:
with open("AF-Q9UPP1-F1-model_v6.pdb", "r") as f:
    pdb_data = f.read()

view = py3Dmol.view(width=900, height=700)
view.addModel(pdb_data, "pdb")

# Show protein cartoon
view.setStyle({'cartoon': {'color': 'lightblue'}})

# Highlight mutation site 643
view.addStyle({'resi': 643}, {'stick': {'color': 'red'}})

view.zoomTo()
view.show()
```
