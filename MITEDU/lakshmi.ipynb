```python
#https://drive.google.com/drive/folders/1lpUnKNA_EKQiPVEe_1tkPuDqcK0BNIGK?usp=drive_link
"""
Normal Distribution

A probability distribution that is symmetric about the mean, showing that data near the mean are more frequent in occurrence than data far from the mean.

Uniform Distribution

A probability distribution in which all outcomes are equally likely.

Standard Distribution

A special normal distribution where the mean is 0 and the standard deviation is 1.

Binomial Distribution

A probability that summarizes the number of trials or observations when each trial has the same probability of attaining one particular value.

Empirical Distribution

A distribution function in which each possible event is assigned a probability derived from experimental observation. It is assumed that the events are independent and the sum of the probabilities is 1.

AUC (Area Under the Curve)

Area under the curve represents the probability that a random positive variable is positioned to the right of a random negative variable. AUC ranges in value from 0 to 1


"""
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, binom
from sklearn.metrics import roc_curve, auc

# Set seed for reproducibility
np.random.seed(42)

# -------------------------------------------------
# 1. Normal Distribution (Mean=2, Std=1.5)
# -------------------------------------------------
plt.figure()

mean = 2
std = 1.5
x = np.linspace(mean - 4*std, mean + 4*std, 1000)
y = norm.pdf(x, mean, std)

plt.plot(x, y)
plt.title("Normal Distribution (Mean=2, Std=1.5)")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.show()


# -------------------------------------------------
# 2. Uniform Distribution (0 to 1)
# -------------------------------------------------
plt.figure()

uniform_data = np.random.uniform(0, 1, 1000)
plt.hist(uniform_data, bins=20, density=True)
plt.title("Uniform Distribution (0 to 1)")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()


# -------------------------------------------------
# 3. Standard Normal Distribution (Mean=0, Std=1)
# -------------------------------------------------
plt.figure()

x_std = np.linspace(-4, 4, 1000)
y_std = norm.pdf(x_std, 0, 1)

plt.plot(x_std, y_std)
plt.title("Standard Normal Distribution (Mean=0, Std=1)")
plt.xlabel("Z")
plt.ylabel("Probability Density")
plt.show()


# -------------------------------------------------
# 4. Binomial Distribution (n=10, p=0.5)
# -------------------------------------------------
plt.figure()

n = 10
p = 0.5
x_binom = np.arange(0, n + 1)
y_binom = binom.pmf(x_binom, n, p)

plt.bar(x_binom, y_binom)
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()


# -------------------------------------------------
# 5. Empirical Distribution
# -------------------------------------------------
plt.figure()

empirical_data = np.random.choice([1, 2, 3, 4, 5], size=1000)
values, counts = np.unique(empirical_data, return_counts=True)
probabilities = counts / len(empirical_data)

plt.bar(values, probabilities)
plt.title("Empirical Distribution")
plt.xlabel("Value")
plt.ylabel("Observed Probability")
plt.show()


# -------------------------------------------------
# 6. ROC Curve & AUC
# -------------------------------------------------
plt.figure()

y_true = np.random.randint(0, 2, 200)
y_scores = np.random.rand(200)

fpr, tpr, thresholds = roc_curve(y_true, y_scores)
roc_auc = auc(fpr, tpr)

plt.plot(fpr, tpr)
plt.title(f"ROC Curve (AUC = {roc_auc:.2f})")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.show()
```
