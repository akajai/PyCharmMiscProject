```python
Common NLP Normalization Techniques (Quick Summary)

1️⃣ Lowercasing
	•	Converts text to lowercase
	•	"Hello" → "hello"
	•	✅ Reduces vocabulary size
	•	❌ Loses capitalization info (bad for NER)

⸻

2️⃣ Punctuation removal
	•	Removes commas, dots, symbols
	•	"Hi!!!" → "Hi"
	•	✅ Useful for classification
	•	❌ Removes emotion/sentiment cues

⸻

3️⃣ Whitespace normalization
	•	Removes extra spaces, tabs, newlines
	•	"NLP   is  fun" → "NLP is fun"

⸻

4️⃣ Stopword removal
	•	Removes common words (is, the, and)
	•	✅ Helps topic modeling, IR
	•	❌ Can hurt meaning in some tasks

⸻

5️⃣ Stemming
	•	Reduces words to root form
	•	"running" → "run"
	•	✅ Fast
	•	❌ Produces non-words

⸻

6️⃣ Lemmatization
	•	Converts words to dictionary form
	•	"better" → "good"
	•	✅ Meaning-preserving
	•	❌ Slower than stemming

⸻

7️⃣ Number normalization
	•	Remove or replace numbers
	•	"2024" → "<num>"
	•	Useful when numbers don’t matter

⸻

8️⃣ Unicode normalization
	•	Normalizes accented characters
	•	"café" → "cafe"

⸻

9️⃣ Case folding
	•	Stronger than lowercasing
	•	Handles Unicode cases (ß → ss)

⸻

🔟 Contraction expansion
	•	"don't" → "do not"
	•	Improves clarity for classical NLP

⸻

1️⃣1️⃣ Spell correction (optional)
	•	"recieve" → "receive"
	•	Risky without context

⸻

1️⃣2️⃣ Emoji & symbol normalization
	•	"😊" → "happy"
	•	Important for sentiment analysis

⸻

⚠️ What NOT to over-normalize
	•	Named entities (NER)
	•	Legal / medical text
	•	LLM prompts
	•	Sentiment intensity (!!!, emojis)
```
