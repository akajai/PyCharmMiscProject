```python
"""Central Limit theorem


Central Limit Theorem (CLT)

🔹 Statement (plain English)

No matter how the population is distributed, if you take large enough random samples, the distribution of the sample mean will be approximately normal.

⸻

🔹 Key conditions
	•	Sample size n ≥ 30 (rule of thumb)
	•	Samples are independent
	•	Population has finite mean and variance

⸻

🔹 What becomes normal?

Not the data — the sample mean.

⸻

🔹 CLT formula

If:
	•	Population mean = μ
	•	Population standard deviation = σ
	•	Sample size = n

Then:
	•	Mean of sample means = μ
	•	Standard deviation of sample means (Standard Error):

SE = \frac{\sigma}{\sqrt{n}}

⸻

🔹 Example

Population distribution: skewed
Take many samples of size n = 50

➡ The means of those samples form a normal distribution

⸻

🔹 Why CLT is important
	•	Allows use of normal distribution
	•	Enables confidence intervals
	•	Enables hypothesis testing

⸻

2️⃣ Estimation

Estimation means using sample data to estimate population parameters (μ, σ, p).

⸻

A️⃣ Point Estimation

A single value estimate.

"""
```
